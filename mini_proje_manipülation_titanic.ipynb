```python
import pandas as pd
import seaborn as sns
df=sns.load_dataset("titanic")
df.head()
```

```python
## Yaşı 18’den büyük olan yolcuları seç büyükten küçüğe sırala
df[df["age"]>18].sort_values("age", ascending=False).head()
```

```python
## Ücreti (fare) 100’den fazla olan yolcuları seç en yüksekten aza
df[df["fare"]>100].sort_values("fare", ascending=False).head()
```

```python
## Kadın ve 1. sınıfta olan yolcuları seç
df[(df["sex"] == "female") & (df["pclass"] == 1)].head()
```

```python
## erkek ve 2. sınıfta olan yolcuları seç
df[(df["sex"]== "male")&( df["pclass"]==2)].head()
```

```python
## age sütununda eksik (NaN) değerleri bul
df["age"].isnull().sum()
```

```python
## En yaşlı 5 kişiyi listele-age sütununa göre azalan sırala, ilk 5 satırı göster.
df.sort_values("age", ascending=False).head()
```

```python
##Hayatta kalan (survived == 1) erkek yolcuları seç sex == 'male' ve survived == 1
df[(df["survived"]==1)& (df["sex"]=="male")].head()
```

```python
## Görev: Ücreti en yüksek ilk 10 yolcuyu getir fare sütununa göre descending sıralayıp ilk 10 yolcuyu seç.
df.sort_values("fare", ascending=False).head(10)
```

```python
## Sadece  sex ve age sütunlarını göster Kolon seçimi yap.
df[["sex","age"]]
```

```python
##Yaşı 30–40 arasında olan yolcuları bul age >= 30 ve age <= 40
df[(df["age"]>=30) & (df["age"])<=40].head(10)
```

```python
## embarked değeri ‘S’ olanları seç embarked == 'S'
df[(df["embarked"]== "S")].head()
```

```python
## Hem 3. sınıfta olup hem hayatta kalamayanları bul pclass == 3 ve survived == 0
df[(df["pclass"]==3)& (df["survived"]==0)].head()
```

```python
## Ücreti 50 ile 150 arasında olan yolcular fare aralık filtresi.
df[(df["fare"]>=20)& (df["fare"]<=150)].head()
```

```python
## Görev: sibsp + parch toplamı 3'ten büyük olanları seç
## sibsp: kardeş/eş sayısı
## parch: anne/baba/çocuk sayısı
## Bu iki sayıyı topluyorsun.
## Eğer toplam 3’ün üzerindeyse, o yolcuyu seçiyorsun.
df[(df["sibsp"] + df["parch"]) > 3].head()
```
